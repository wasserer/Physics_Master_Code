# Plot the temperature Profile:

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
temperature_log = ""  # Enter your file path/folder path in this place
```

```python
df = pd.read_csv(temperature_log)
```

```python
df
```

```python
time_s = df["Elapsed_s"]
temperature = df["Temp_A_C"]
set_temp = df["Setpoint_C"]
```

```python
temperature_difference = df["Setpoint_C"] - df["Temp_A_C"]
```

```python
time_h = np.array(time_s / 3600)
temperature = np.array(temperature)
```

```python
print(time_h)
print(temperature)
```

```python
plt.figure(figsize = (10, 4), dpi = 300)

plt.plot(time_h, set_temp, alpha = 1, color = "#ED7D31", label = "Set Temperature")
plt.plot(time_h, temperature, alpha = 0.8, color = "#5B8CBE", label = "Operation Temperature")
#plt.xlim(0, 3.2)
plt.ylim(-115, 80)
plt.xlabel("Time [h]")
plt.ylabel("Temperature [$^{\circ}$C]")
plt.legend()
plt.grid()
```

```python
plt.figure(figsize = (5, 4), dpi = 300)

# Shade eclipse phase where setpoint is below 0C; adjust threshold if needed.
eclipse_mask = (set_temp < 0).to_numpy() if hasattr(set_temp, "to_numpy") else (set_temp < 0)
if np.any(eclipse_mask):
    edges = np.diff(eclipse_mask.astype(int))
    start_idx = np.where(edges == 1)[0] + 1
    end_idx = np.where(edges == -1)[0] + 1
    if eclipse_mask[0]:
        start_idx = np.r_[0, start_idx]
    if eclipse_mask[-1]:
        end_idx = np.r_[end_idx, len(eclipse_mask) - 1]
    for s_idx, e_idx in zip(start_idx, end_idx):
        time_vals = time_h.to_numpy() if hasattr(time_h, "to_numpy") else time_h
        plt.axvspan(time_vals[s_idx], time_vals[e_idx], color="lightgray", alpha=0.4, zorder=0)

plt.plot(time_h, set_temp, alpha = 1, color = "#ED7D31")
#plt.plot(time_h, temperature, alpha = 0.8, color = "#5B8CBE", label = "Operation Temperature")
plt.xlim(0, 1.52)
#plt.ylim(-80, 80)
plt.xlabel("Time [h]")
plt.ylabel("Temperature [$^{\circ}$C]")
#plt.legend()
#plt.grid()
```

```python
plt.figure(figsize = (10, 4), dpi = 300)
plt.plot(time_h, temperature_difference, alpha = 0.8, label = "Operation Temperature")
plt.xlim(40, 57)
plt.ylim(-10, 10)
plt.xlabel("Time [h]")
plt.ylabel("Temperature [$^{\circ}$C]")
plt.legend()
plt.grid()
```

```python
plt.figure(figsize = (10, 4), dpi = 300)
plt.plot(time_h, temperature_difference, alpha = 0.8, label = "Operation Temperature")
plt.xlim(9, 11)
plt.ylim(-10, 10)
plt.xlabel("Time [hour]")
plt.ylabel("Temperature [$^{\circ}$C]")
plt.legend()
plt.grid()
```

```python
fig, ax1 = plt.subplots(figsize=(8, 4), dpi = 300)

# Plot the first y-axis data
color1 = 'tab:blue'
ax1.set_xlabel('Time [h]')
ax1.set_ylim(-120, 80)
ax1.set_ylabel("Temperature [$^{\circ}$C]", color=color1)
ax1.plot(time_h, set_temp, color=color1, label='Set Temperature')
ax1.tick_params(axis='y', labelcolor=color1)

# Create the second y-axis sharing the same x-axis
ax2 = ax1.twinx()
color2 = 'tab:red'
ax2.set_ylabel('Temperature Difference [$^{\circ}$C]', color=color2)
ax2.plot(time_h, temperature_difference, color=color2, label='Temperature Difference')
ax2.tick_params(axis='y', labelcolor=color2)

#ax2.set_xlim(55.5, 70)
ax2.set_xlim(0, 12.25)
ax2.set_ylim(-15, 15)
# Optional: combine legends
lines_1, labels_1 = ax1.get_legend_handles_labels()
lines_2, labels_2 = ax2.get_legend_handles_labels()
ax1.legend(lines_1 + lines_2, labels_1 + labels_2, loc='lower right')

plt.tight_layout()
```
