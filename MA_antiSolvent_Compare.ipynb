# Anti-solvent Comparison

This notebook compares Voc, Isc, FF, and PCE across four processing conditions: R3000_20s, R3000_25s, R5000_25s, R5000_30s.

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from plot_module.solsim_analyzer import solarSimulator

sns.set(style='whitegrid', context='talk')

# Helper to load metrics from a folder of .dat files
def load_metrics(folder: str):
    """Return dict of lists for Isc, Voc, FF, PCE from a folder."""
    if not os.path.isdir(folder):
        raise FileNotFoundError(f'Folder not found: {folder}')
    analyzer = solarSimulator(folderPath=folder)
    analyzer.loadFolderData()
    return {
        'Isc': list(analyzer.Isc),
        'Voc': list(analyzer.Voc),
        'FF':  list(analyzer.FF),
        'PCE': list(analyzer.PCE),
    }
```

```python
#CODEX: CHANGE THE CODE FROM HERE!
#Compare the diffferent parameters(in average), and plot them in box plot and compare their average value.
# Build a long-form DataFrame for boxplots
all_rows = []
for group, folder in GROUPS.items():
    if not folder:
        continue
    try:
        metrics = load_metrics(folder)
    except Exception as e:
        print(f'Skipping {group}: {e}')
        continue
    for metric_name, values in metrics.items():
        for val in values:
            all_rows.append({'Group': group, 'Metric': metric_name, 'Value': val})

if not all_rows:
    raise RuntimeError('No data loaded. Fill the folder paths and re-run.')

long_df = pd.DataFrame(all_rows)

# 1) Convert FF to percentage
long_df.loc[long_df['Metric'] == 'FF', 'Value'] = \
    long_df.loc[long_df['Metric'] == 'FF', 'Value'] * 100.0

# Optional: compute and display group mean ± SEM (after conversion)
print('Metrics legend: Voc = open-circuit voltage (V), Isc = short-circuit current density (mA/cm^2), FF = fill factor (%), PCE = power conversion efficiency (%)')
group_stats = long_df.groupby(['Metric','Group'])['Value'].agg(mean='mean', std='std', count='count')
def _format_mean_sem(row):
    mean = row['mean']
    count = row['count']
    std = row['std']
    if count <= 1 or pd.isna(std):
        return f"{mean:.3f} ± n/a"
    sem = std / (count ** 0.5)
    return f"{mean:.3f} ± {sem:.3f}"
formatted_stats = group_stats.apply(_format_mean_sem, axis=1).unstack()
print('Group mean ± SEM:')
print(formatted_stats)
# 2) Plot with brighter colors: one boxplot per metric (2x2 grid)
metrics_order = ['Voc', 'Isc', 'FF', 'PCE']
fig, axes = plt.subplots(2, 2, figsize=(14, 10), dpi=150)
axes = axes.ravel()

# Define per-group colors (brighter orange for R3000_25s and brighter red for R5000_30s)
groups_order = list(GROUPS.keys())
group_palette = {
    'R3000_20s': '#EB1615',  # blue
    'R3000_25s': '#EBBD15',  # brighter saturated orange
    'R5000_25s': '#35B85B',  # green
    'R5000_30s': '#3C4CFF',  # brighter saturated red
}
palette = [group_palette[g] for g in groups_order]

for ax, metric in zip(axes, metrics_order):
    sub = long_df[long_df['Metric'] == metric]
    sns.boxplot(data=sub, x='Group', y='Value', ax=ax, order=groups_order, palette=palette)
    sns.stripplot(data=sub, x='Group', y='Value', ax=ax, order=groups_order, color='black', size=3, jitter=True, alpha=0.6)
    ax.set_xlabel('')
    if metric == 'FF':
        ax.set_ylabel('FF (%)')
        ax.set_ylim(0, 100)
    else:
        ax.set_ylabel(metric)
    ax.set_title(f'{metric} comparison')
    ax.grid(True, axis='y', alpha=0.3)
    for tick in ax.get_xticklabels():
        tick.set_rotation(20)
        tick.set_ha('right')

plt.tight_layout()
plt.show()

# 3) Plot IV curves for best PCE in each group
fig, ax = plt.subplots(figsize=(8, 6), dpi=150)
# Use the same group-specific colors for IV curves
iv_colors = group_palette
for (idx_g, (group, folder)) in enumerate(GROUPS.items()):
    if not folder:
        continue
    try:
        analyzer = solarSimulator(folderPath=folder)
        analyzer.loadFolderData()
        if not analyzer.PCE:
            continue
        best_idx = max(range(len(analyzer.PCE)), key=lambda i: analyzer.PCE[i])
        v = analyzer.voltages[best_idx]
        j = analyzer.currents[best_idx]
        ax.plot(v, j, label=f'{group}', color=iv_colors.get(group, None), linewidth=2.2)
    except Exception as e:
        print(f'Skipping IV for {group}: {e}')

ax.set_xlabel('Voltage (V)')
ax.set_ylabel('Current density (mA/cm^2)')
#ax.set_title('IV-Curves of the solar cells with')
# Requested axis limits
ax.set_ylim(-1, 20)
ax.set_xlim(-0.95, )
ax.grid(True, alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()
```
