```python
#num = 1471
for num in range (1300, 1302):
    plt.figure(dpi = 150)
    plt.plot(spectra.eV, spectra.valueTaucs[num], label='Tauc Plot at temperature {:.1f} Degree'.format(spectra.temperatures[num]))
    plt.plot(spectra.eV, spectra.tauc_slope_b[num] + spectra.tauc_slope[num] * spectra.eV, label=r'Fit Line, band gap = {:.3f} eV'.format(spectra.bandGap[num]))
    plt.xlim(1.45, 2)
    plt.ylim(0,2)
    plt.xlabel('Energy (eV)')
    plt.ylabel('Tauc Plot (Absorbance * photon energy)^0.5')
    plt.title('Tauc Plot for Band Gap Calculation')
    plt.legend()
    plt.grid()
```
