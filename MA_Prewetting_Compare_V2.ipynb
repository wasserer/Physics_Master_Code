```python
wet = solarSimulator(folderPath=folderPath_With, refCurrent=32)
dry = solarSimulator(folderPath=folderPath_Without, refCurrent=32)
wet.loadFolderData()
dry.loadFolderData()
```

```python
for i in range (0, len(wet.PCE)):
    print("Name:", wet.labels[i])
    print("Voc:", wet.Voc[i])
    print("Isc:", wet.Isc[i])
    print("FF:", wet.FF[i])
    print("PCE:", wet.PCE[i])

for i in range (0, len(dry.PCE)):
    print("Name:", dry.labels[i])
    print("Voc:", dry.Voc[i])
    print("Isc:", dry.Isc[i])
    print("FF:", dry.FF[i])
    print("PCE:", dry.PCE[i])
```

```python
#Codex: Write your code here:
# Box plots: With vs Without Prewetting for Voc, Isc, FF, PCE
metrics = ['Voc', 'Isc', 'FF', 'PCE']
records = []
for m in metrics:
    w_vals = getattr(wet, m)
    d_vals = getattr(dry, m)
    # Flatten possible numpy arrays and ensure floats
    for v in list(w_vals):
        records.append({'Metric': m, 'Group': 'With Prewetting', 'Value': float(v)})
    for v in list(d_vals):
        records.append({'Metric': m, 'Group': 'Without Prewetting', 'Value': float(v)})

df = pd.DataFrame(records)
palette = {'With Prewetting': '#0072BD', 'Without Prewetting': '#FF1744'}

fig, axes = plt.subplots(2, 2, figsize=(12, 10), dpi=150)
for i, m in enumerate(metrics):
    ax = axes[i // 2][i % 2]
    data_m = df[df['Metric'] == m]
    sns.boxplot(data=data_m, x='Group', y='Value', ax=ax, order=['With Prewetting', 'Without Prewetting'], palette=palette)
    sns.stripplot(data=data_m, x='Group', y='Value', ax=ax, order=['With Prewetting', 'Without Prewetting'], color='black', size=4, jitter=True, alpha=0.7)
    # Horizontal grid lines
    ax.yaxis.grid(True, linestyle='-', alpha=0.4)
    ax.set_axisbelow(True)
    ax.set_title(m)
    if m == 'Isc':
        ax.set_ylabel('Isc [mA/cm2]')
    elif m == 'Voc':
        ax.set_ylabel('Voc [V]')
    elif m == 'FF':
        ax.set_ylabel('FF')
    elif m == 'PCE':
        ax.set_ylabel('PCE [%]')
    ax.set_xlabel('')

#fig.suptitle('With vs Without Prewetting: Voc, Isc, FF, PCE', y=0.98)
plt.tight_layout(rect=[0, 0, 1, 0.96])
```
