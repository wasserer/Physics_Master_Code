# Compare the Perovskite with and without prewetting:

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from plot_module.solsim_analyzer import solarSimulator

#sns.set(style='whitegrid', context='talk')
```

### Load the folders:

```python
folderPath_With = ""  # Enter your file path/folder path in this place
folderPath_Without = ""  # Enter your file path/folder path in this place
```

```python
wet = solarSimulator(folderPath=folderPath_With, refCurrent=32)
dry = solarSimulator(folderPath=folderPath_Without, refCurrent=32)
wet.loadFolderData()
dry.loadFolderData()
```

```python
for i in range (0, len(wet.PCE)):
    print("Name:", wet.labels[i])
    print("Voc:", wet.Voc[i])
    print("Isc:", wet.Isc[i])
    print("FF:", wet.FF[i])
    print("PCE:", wet.PCE[i])

for i in range (0, len(dry.PCE)):
    print("Name:", dry.labels[i])
    print("Voc:", dry.Voc[i])
    print("Isc:", dry.Isc[i])
    print("FF:", dry.FF[i])
    print("PCE:", dry.PCE[i])
```

```python
# Box plots: With vs Without Prewetting for Voc, Isc, FF, PCE
metrics = ['Voc', 'Isc', 'FF', 'PCE']
records = []
for m in metrics:
    w_vals = getattr(wet, m)
    d_vals = getattr(dry, m)
    # Flatten possible numpy arrays and ensure floats
    for v in list(w_vals):
        records.append({'Metric': m, 'Group': 'Prewetting', 'Value': float(v)})
    for v in list(d_vals):
        records.append({'Metric': m, 'Group': 'Without Prewetting', 'Value': float(v)})

df = pd.DataFrame(records)
palette = {'Prewetting': '#3C4CFF', 'Without Prewetting': '#EBBD15'}#['#3C4CFF', '#EBBD15']

fig, axes = plt.subplots(2, 2, figsize=(8, 6), dpi=300)
for i, m in enumerate(metrics):
    ax = axes[i // 2][i % 2]
    data_m = df[df['Metric'] == m]
    sns.boxplot(data=data_m, x='Group', y='Value', ax=ax, order=['Prewetting', 'Without Prewetting'], palette=palette)
    sns.stripplot(data=data_m, x='Group', y='Value', ax=ax, order=['Prewetting', 'Without Prewetting'], color='black', size=4, jitter=True, alpha=0.7)
    # Horizontal grid lines
    ax.yaxis.grid(True, linestyle='-', alpha=0.4)
    ax.set_axisbelow(True)
    ax.set_title(m)
    if m == 'Isc':
        ax.set_ylabel('Isc [mA/cm2]')
    elif m == 'Voc':
        ax.set_ylabel('Voc [V]')
    elif m == 'FF':
        ax.set_ylabel('FF')
    elif m == 'PCE':
        ax.set_ylabel('PCE [%]')
    ax.set_xlabel('')

#fig.suptitle('With vs Without Prewetting: Voc, Isc, FF, PCE', y=0.98)
plt.tight_layout(rect=[0, 0, 1, 0.96])
```

```python
# Box plots: With vs Without Prewetting for Voc, Isc, FF, PCE
metrics = ['Voc', 'Isc', 'FF', 'PCE']
records = []

for m in metrics:
    w_vals = getattr(wet, m)
    d_vals = getattr(dry, m)
    for v in list(w_vals):
        records.append({'Metric': m, 'Group': 'Prewetting', 'Value': float(v)})
    for v in list(d_vals):
        records.append({'Metric': m, 'Group': 'Without Prewetting', 'Value': float(v)})

df = pd.DataFrame(records)
palette = {'Prewetting': '#3C4CFF', 'Without Prewetting': '#EBBD15'}

fig, axes = plt.subplots(2, 2, figsize=(8, 6), dpi=300)

for i, m in enumerate(metrics):
    ax = axes[i // 2][i % 2]
    data_m = df[df['Metric'] == m]

    sns.boxplot(
        data=data_m, x='Group', y='Value',
        ax=ax, order=['Prewetting', 'Without Prewetting'], palette=palette
    )
    sns.stripplot(
        data=data_m, x='Group', y='Value',
        ax=ax, order=['Prewetting', 'Without Prewetting'],
        color='black', size=4, jitter=True, alpha=0.7
    )

    # --- 改成灰色边框 ---
    for artist in ax.artists:
        artist.set_edgecolor("gray")
    for line in ax.lines:
        line.set_color("gray")
    # ----------------------

    # Horizontal grid lines
    ax.yaxis.grid(True, linestyle='-', alpha=0.4)
    ax.set_axisbelow(True)
    ax.set_title(m)

    if m == 'Isc':
        ax.set_ylabel('Isc [mA/cm2]')
    elif m == 'Voc':
        ax.set_ylabel('Voc [V]')
    elif m == 'FF':
        ax.set_ylabel('FF')
    elif m == 'PCE':
        ax.set_ylabel('PCE [%]')
    ax.set_xlabel('')

plt.tight_layout(rect=[0, 0, 1, 0.96])
```

```python
# Calculate average and standard error (std/sqrt(n)) for each metric/group (boxplot-style outliers removed)
def exclude_outliers(group):
    values = group['Value']
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return group[(values >= lower) & (values <= upper)]

filtered = (
    df.groupby(['Metric', 'Group'], group_keys=False)
      .apply(exclude_outliers)
)

summary = (
    filtered.groupby(['Metric', 'Group'])['Value']
      .agg(mean='mean', std='std', count='count')
      .reset_index()
)
summary['error'] = summary['std'].fillna(0) / np.sqrt(summary['count'].clip(lower=1))

for _, row in summary.iterrows():
    metric = row['Metric']
    group = row['Group']
    mean = row['mean']
    error = row['error']
    count = int(row['count'])
    print(f"{metric} ({group}) -> mean: {mean:.3f}, error: {error:.3f} (n={count})")

```
