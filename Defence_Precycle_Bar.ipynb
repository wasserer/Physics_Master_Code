# Do a bar plot for the 6 pixels before the thermal cycling:

```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
colors = (
    '#1f77b4',
    '#ff7f0e',
    '#2ca02c',
    '#d62728',
    '#9467bd',
    '#8c564b',
)
```

```python
px1 = [15.474, 0.837, 74.087, 14.387]
px2 = [16.239, 0.907, 77.372, 17.079]
px3 = [14.089, 0.900, 71.915, 13.681]
px4 = [16.589, 0.938, 74.166, 17.297]
px5 = [15.473, 0.862, 45.480, 09.098]
px6 = [16.953, 0.868, 79.987, 17.639]
```

```python
for i in range(len(px1)):
    px2[i] = px2[i] / px6[i]
    px3[i] = px3[i] / px6[i]
    px4[i] = px4[i] / px6[i]
    px5[i] = px5[i] / px6[i]
    px1[i] = px1[i] / px6[i]
    px6[i] = px6[i] / px6[i]
```

```python
array = np.array([px1, px2, px3, px4, px5, px6])
data = array

labels = ["I$_{sc}$ [mA/cm$^2$]", "V$_{oc}$ [V]", "FF", "PCE [%]"]
```

```python
print(data.shape[0])
```

```python
n_groups = data.shape[0]      # 6
n_bars = data.shape[1]
x = np.arange(n_groups)
bar_width = 0.18
```

```python
fig, ax = plt.subplots(figsize=(5, 3), dpi=300)

for i in range(n_bars):
    ax.bar(
        x + (i - (n_bars - 1) / 2) * bar_width,
        data[:, i],
        width=bar_width,
        color=colors[i],
        label=labels[i],
        edgecolor='black',
    )

# ======================
# 4. 美化
# ======================
ax.set_xticks(x)
ax.set_xticklabels([f'Pixel {i+1}' for i in range(n_groups)])
#ax.set_xlim(0, 9.2)
#ax.set_ylim(0, 1.7)
ax.set_ylabel('Value (normalized)')
#ax.set_title('Grouped Bar Chart with Fixed Colors')
#ax.legend()


plt.tight_layout()
#plt.grid()
plt.legend()
plt.show()
```
